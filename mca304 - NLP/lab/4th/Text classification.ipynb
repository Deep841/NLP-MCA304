```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

# ----------------------
# 1. Load dataset
# ----------------------
# IMDB movie reviews (25k train, 25k test)




max_features = 10000   # only keep top 10k words
(x_train, y_train), (x_test, y_test) = keras.datasets.imdb.load_data(num_words=max_features)
```

```python

max_len = 200          # cut reviews after 200 words
```

```python

# ----------------------
# 2. Preprocess text
# ----------------------
# Pad sequences so all reviews are the same length
x_train = keras.preprocessing.sequence.pad_sequences(x_train, maxlen=max_len)
x_test = keras.preprocessing.sequence.pad_sequences(x_test, maxlen=max_len)
```

Makes all sequences the same length (maxlen=200 here).

Short reviews → are padded with zeros (default) at the beginning.

Long reviews → are truncated to the last 200 words.

 [   0,   0,  12, 789,  34],    # padded with zeros

 [  98,  65,  34,  21, 456]]    # truncated to last 5


```python


# ----------------------
# 3. Build Model
# ----------------------
model = keras.Sequential([
    layers.Embedding(input_dim=max_features, output_dim=128, input_length=max_len),
    layers.Bidirectional(layers.LSTM(64, return_sequences=False)),
    layers.Dense(64, activation="relu"),
    layers.Dense(1, activation="sigmoid")  # binary classification
])
```

Input:  (batch_size, max_len)

Output: (batch_size, max_len, output_dim)

embeddings learnt using backpropagation as a part of learning process


Each review (sequence) becomes a matrix of shape (max_len, 128).

```python

# ----------------------
# 4. Compile Model
# ----------------------
model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)
```

```python


# ----------------------
# 5. Train
# ----------------------
history = model.fit(
    x_train, y_train,
    epochs=5,
    batch_size=64,
    validation_split=0.2
)
```

```python

# ----------------------
# 6. Evaluate
# ----------------------
loss, acc = model.evaluate(x_test, y_test)
print(f"Test Accuracy: {acc:.4f}")
```
