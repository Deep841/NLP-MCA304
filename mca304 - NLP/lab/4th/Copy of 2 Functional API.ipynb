Suppose you are predicting disease risk using:

Tabular features (age, weight, blood pressure → vector).

Image features (X-ray image → matrix).

```python
from tensorflow.keras import Model, layers, Input

# Input 1: tabular data (10 features)
tabular_input = Input(shape=(10,), name="tabular_input")
x1 = layers.Dense(16, activation='relu')(tabular_input)

# Input 2: image data (64x64 RGB)
image_input = Input(shape=(64, 64, 3), name="image_input")
x2 = layers.Conv2D(32, (3,3), activation='relu')(image_input)
x2 = layers.Flatten()(x2)

# Merge both inputs
merged = layers.concatenate([x1, x2])

# Final output
output = layers.Dense(1, activation='sigmoid', name="output")(merged)

# Define model
model = Model(inputs=[tabular_input, image_input], outputs=output)
```

Multiple Outputs

Suppose you want a model that:

Predicts age (regression output).

Predicts gender (classification output).

```python
# Input: face image
inputs = Input(shape=(64, 64, 3), name="image_input")

x = layers.Conv2D(32, (3,3), activation='relu')(inputs)
x = layers.Flatten()(x)

# Output 1: Age (regression)
age_output = layers.Dense(1, name="age_output")(x)

# Output 2: Gender (binary classification)
gender_output = layers.Dense(1, activation='sigmoid', name="gender_output")(x)

# Define model with multiple outputs
model = Model(inputs=inputs, outputs=[age_output, gender_output])
```

Multiple Inputs + Multiple Outputs

Example:

Inputs = (tabular + image).

Outputs = (disease risk + treatment type).

```python
# Tabular input
tabular_input = Input(shape=(10,), name="tabular_input")
x1 = layers.Dense(16, activation='relu')(tabular_input)

# Image input
image_input = Input(shape=(64, 64, 3), name="image_input")
x2 = layers.Conv2D(32, (3,3), activation='relu')(image_input)
x2 = layers.Flatten()(x2)

# Merge inputs
merged = layers.concatenate([x1, x2])

# Output 1: Disease risk
risk_output = layers.Dense(1, activation='sigmoid', name="risk_output")(merged)

# Output 2: Treatment type (multi-class classification)
treatment_output = layers.Dense(3, activation='softmax', name="treatment_output")(merged)

# Define model
model = Model(inputs=[tabular_input, image_input], outputs=[risk_output, treatment_output])
```
