```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

# Load MNIST dataset
(x_train, y_train), (x_test, y_test) = keras.datasets.mnist.load_data()

# Normalize and reshape
x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0

# Flatten images for dense NN
x_train = x_train.reshape(-1, 28*28)
x_test = x_test.reshape(-1, 28*28)
```

tf.data.Dataset.from_tensor_slices((x_train, y_train))

Creates a TensorFlow Dataset object from the training data.

It pairs each x_train[i] with y_train[i], like (image, label) pairs.

3. .shuffle(10000)

Shuffles the dataset to ensure batches are randomized (important for training stability).

The number 10000 is the buffer size → how many samples TensorFlow loads into memory for shuffling.

A large buffer means better shuffling (more randomness), but uses more memory.

```python


batch_size = 64
train_dataset = tf.data.Dataset.from_tensor_slices((x_train, y_train)).shuffle(10000).batch(batch_size)
test_dataset = tf.data.Dataset.from_tensor_slices((x_test, y_test)).batch(batch_size)

# ----------------------
# Method 1: Sequential API
# ----------------------
model_seq = keras.Sequential([
    layers.Input(shape=(28*28,)),
    layers.Dense(512, activation="relu"),
    layers.Dense(512, activation="relu"),
    layers.Dense(10)  # logits
])

model_seq.compile(
    optimizer=keras.optimizers.Adam(learning_rate=3e-4),
    loss=keras.losses.SparseCategoricalCrossentropy(from_logits=True),
    metrics=["accuracy"]
)

print("Training Sequential model...")
model_seq.fit(train_dataset, epochs=10, validation_data=test_dataset)
print("Evaluating Sequential model...")
model_seq.evaluate(test_dataset)

# ----------------------
# Method 2: Functional API
# ----------------------
inputs = keras.Input(shape=(28*28,))
x = layers.Dense(512, activation="relu")(inputs)
x = layers.Dense(512, activation="relu")(x)
outputs = layers.Dense(10)(x)  # logits

model_func = keras.Model(inputs=inputs, outputs=outputs)

model_func.compile(
    optimizer=keras.optimizers.Adam(learning_rate=3e-4),
    loss=keras.losses.SparseCategoricalCrossentropy(from_logits=True),
    metrics=["accuracy"]
)

print("Training Functional API model...")
model_func.fit(train_dataset, epochs=10, validation_data=test_dataset)
print("Evaluating Functional API model...")
model_func.evaluate(test_dataset)
```
