```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras.datasets import mnist
```

lists all the physical GPU devices available on your machine.

```python
physical_devices = tf.config.list_physical_devices("GPU")
```

This command enables dynamic memory growth for the first GPU (physical_devices[0]). By default, TensorFlow pre-allocates most GPU memory, which can be inefficient when running multiple models or processes. Setting memory_growth to True ensures TensorFlow only uses GPU memory as needed, making it more efficient for shared GPU use.

```python
tf.config.experimental.set_memory_growth(physical_devices[0], True)
```

```python
(x_train0, y_train0), (x_test0, y_test0) = mnist.load_data()
```

```python
x_train = x_train0.reshape(-1, 28 * 28).astype("float32") / 255.0
x_test = x_test0.reshape(-1, 28 * 28).astype("float32") / 255.0
```

# Sequential API (Very convenient, not very flexible)
say one input,one output

```python

model = keras.Sequential(
    [
        keras.Input(shape=(28 * 28,)),
        layers.Dense(512, activation="relu",name='hidden_1'),
        layers.Dense(256, activation="relu",name='hidden_2'),
        #layers.Dense(10,name='Output')
        layers.Dense(10,activation='softmax',name='output')

    ]
)



#No activation function is specified in last, meaning this layer outputs raw logits (unnormalized scores).
```

```python
model.summary()
```

To configure the model for training. It specifies the loss function, optimizer, and metrics that will be used during the training process.

```python
model1 = keras.Sequential(
    [

        layers.Dense(512, activation="relu",name='hidden_1'),
        layers.Dense(256, activation="relu",name='hidden_2'),
        #layers.Dense(10,name='Output')
        layers.Dense(10,activation='softmax',name='output')

    ]
)
model1.summary()
```

```python
model.compile(
    loss=keras.losses.SparseCategoricalCrossentropy(from_logits=False),      # means the output is already in probability form (softmax has been applied).
    optimizer=keras.optimizers.Adam(learning_rate=0.001),
    metrics=["accuracy"]
)
```

```python
model.fit(x_train, y_train0, batch_size=32, epochs=5, verbose=True)
```

```python
model.evaluate(x_test, y_test, batch_size=32, verbose=True)
```

Larger batch sizes: Faster evaluation but require more memory.

Smaller batch sizes:
Slower evaluation but use less memory.

# Functional API (A bit more flexible)

```python

inputs = keras.Input(shape=(784,))
x = layers.Dense(512, activation="relu", name="first_layer")(inputs)
x = layers.Dense(256, activation="relu", name="second_layer")(x)
outputs = layers.Dense(10, activation="softmax")(x)
model = keras.Model(inputs=inputs, outputs=outputs)
```

```python
model.compile(
    loss=keras.losses.SparseCategoricalCrossentropy(from_logits=False),
    optimizer=keras.optimizers.Adam(learning_rate=0.001),
    metrics=["accuracy"],
)

```
