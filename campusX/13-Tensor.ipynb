Tensors 

```python
import numpy as np
```

```python
# 0-D Tensor (Scalar)
scalar = np.array(7)
print("Scalar (0D):", scalar)
print("Shape:", scalar.shape)
print("Dimensions:", scalar.ndim)
```

```python
#1-D Tensor (Vector)
vector = np.array([31,1,2,3])
print("Vector (1D):\n",vector)
print("Shape:", vector.shape)
print("Dimensions:", vector.ndim)
```

```python
#2-D Tensor (Matrix)
matrix = np.array([[31,31],[31,31],[31,31]])
print("Matrix (2D):\n", matrix)
print("Shape:", matrix.shape)
print("Dimensions:", matrix.ndim)
```

```python
#3-D Tensor (Cube like structure) : 2x2x3
tensor3d = np.array([
    [[31,31,31], [31,31,31]],
    [[31,31,31], [31,31,31]]
    ])
print("3D:\n", tensor3d)
print("Shape:", tensor3d.shape)
print("Dimensions:", tensor3d.ndim)
```

```python
# n-D Tensor (4D example, often used for images: batch, height, width, channels)
tensor4d = np.random.rand(2, 3, 4, 5)  # shape = (2 batches, 3 height, 4 width, 5 channels)
print("4D Tensor shape:", tensor4d.shape)
```
